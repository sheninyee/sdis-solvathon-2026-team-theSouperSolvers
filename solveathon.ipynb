```python
data_path = '/content/drive/My Drive/grants-search-202608182008.csv'
df = pd.read_csv(data_path)

df.info()
df.head(60)
```

```python
columns_needed = [
    'opportunity_number',
    'opportunity_title',
    'category',
    'funding_categories',
    'summary_description',
    'applicant_types',
    'applicant_eligibility_description',
    'close_date',
    'award_floor',
    'award_ceiling',
    'estimated_total_program_funding',
    'is_cost_sharing',
    'is_forecast',
    'url'
]
grants = df[columns_needed].copy()

grants.head(20)
```

```python
# Categorizes applicant types eligibility
applicant_types = (
    df["applicant_types"]
    .fillna("")
    .str.split(";")
    .explode()
)

applicant_summary = (
    applicant_types
    .value_counts()
    .reset_index()
)

applicant_summary.columns = [
    "Applicant Type",
    "Number of Grants"
]

applicant_summary
```

```python
# Hard filter 1: Eligibility for state government (NCDOT)
grants["eligibility_status"] = (
    grants["applicant_types"]
    .fillna("")
    .str.split(";")
    .apply(
        lambda x: (
            "Potentially eligible"
            if "state_governments" in x
            else "Not eligible"
        )
    )
)
eligible = grants[
    grants["eligibility_status"] == "Potentially eligible"
].copy()

eligible["funding_categories"] = (
    eligible["funding_categories"]
    .fillna("")
    .str.split(";")
)

category_summary = (
    eligible["funding_categories"]
    .explode()
    .value_counts()
    .reset_index()
)

category_summary.columns = [
    "Funding Category",
    "Number of Grants"
]

category_summary
```

```python
# Recheck individual grants
grants[
    [
        "opportunity_title",
        "applicant_types",
        "eligibility_status"
    ]
].head(20)
```

```python
# Split data by categories

grants["funding_categories"] = (
    grants["funding_categories"]
    .fillna("")
    .str.split(";")
)

grants[["opportunity_title", "funding_categories"]].head()
```

```python
eligible = grants[
    grants["eligibility_status"] == "Potentially eligible"
].copy()
```

```python
all_categories = (
    eligible["funding_categories"]
    .explode()
)

all_categories.value_counts()
```

```python
# Convert close_date to datetime
eligible["close_date"] = pd.to_datetime(
    eligible["close_date"],
    errors="coerce"
)
today = pd.Timestamp("2026-09-24")

eligible["deadline_status"] = eligible["close_date"].apply(
    lambda x:
        "Missing deadline" if pd.isna(x)
        else "Future" if x >= today
        else "Passed"
)

eligible["deadline_status"].value_counts()
```

```python
# Check how many grants have missing deadlines are forecast grants
eligible[
    eligible["close_date"].isna()
]["is_forecast"].value_counts(dropna=False)
```

```python
# Hard filter 2: Grants that deadlines have already passed
eligible["close_date"] = pd.to_datetime(
    eligible["close_date"],
    errors="coerce"
)

reference_date = pd.Timestamp("2026-09-24")

eligible_open = eligible[
    eligible["close_date"].isna() |
    (eligible["close_date"] >= reference_date)
].copy()
```

```python
# Calculate award range
eligible_open["award_range"] = (
    eligible_open["award_ceiling"] - eligible_open["award_floor"]
)
eligible_open[
    ["opportunity_title", "award_floor", "award_ceiling", "award_range"]
].head(20)
```

```python
# Scoring: Funding categories
funding_category_scores = {
    "transportation": 3,
    "infrastructure_investment_and_jobs_act": 3,

    "disaster_prevention_and_relief": 2,
    "community_development": 2,
    "regional_development": 2,
    "science_technology_and_other_research_and_development": 2,
    "information_and_statistics": 2,

    "environment": 1,
    "energy": 1,
    "employment_labor_and_training": 1
}

def score_funding_category(categories):
    scores = []

    for category in categories:
        if category in funding_category_scores:
            scores.append(funding_category_scores[category])
        else:
            scores.append(0)

    return max(scores) if scores else 0

eligible_open["funding_category_score"] = (
    eligible_open["funding_categories"]
    .apply(score_funding_category)
)
```

```python
# Scoring: Categories ((discretionary, mandatory, earmark, other)
def score_category(category):

    if category == "discretionary":
        return 3

    elif category == "mandatory":
        return 2

    elif category == "earmark":
        return 1

    else:
        return 0

eligible_open["category_score"] = (
    eligible_open["category"]
    .fillna("")
    .str.lower()
    .apply(score_category)
)
```

```python
# Scoring: Cost sharing?
def score_cost_sharing(cost_sharing):

    if cost_sharing == True:
        return -5

    else:
        return 0

eligible_open["cost_sharing_score"] = (
    eligible_open["is_cost_sharing"]
    .apply(score_cost_sharing)
)
```

```python
# Scoring: Forecast?
def score_forecast(is_forecast):

    if is_forecast == True:
        return -1

    else:
        return 0

eligible_open["forecast_score"] = (
    eligible_open["is_forecast"]
    .apply(score_forecast)
)
```

```python
# Scoring: Close date
reference_date = pd.Timestamp("2026-09-25")

eligible_open["days_until_close"] = (
    eligible_open["close_date"] - reference_date
).dt.days
```

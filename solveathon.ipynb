```python
# Scoring: Funding categories
funding_category_scores = {
    "transportation": 10,
    "infrastructure_investment_and_jobs_act": 10,

    "disaster_prevention_and_relief": 3,
    "community_development": 2,
    "regional_development": 2,
    "science_technology_and_other_research_and_development": 2,
    "information_and_statistics": 2,

    "environment": 1,
    "energy": 1,
    "employment_labor_and_training": 1
}

def score_funding_category(categories):
    scores = []

    for category in categories:
        if category in funding_category_scores:
            scores.append(funding_category_scores[category])
        else:
            scores.append(0)

    return max(scores) if scores else 0

eligible_final["funding_category_score"] = (
    eligible_final["funding_categories"]
    .apply(score_funding_category)
)
```

```python
# Scoring: Categories ((discretionary, mandatory, earmark, other)
def score_category(category):

    if category == "discretionary":
        return 3

    elif category == "mandatory":
        return 2

    elif category == "earmark":
        return 1

    else:
        return 0

eligible_final["category_score"] = (
    eligible_final["category"]
    .fillna("")
    .str.lower()
    .apply(score_category)
)
```

```python
# Scoring: Close date
reference_date = pd.Timestamp("2026-09-25")

eligible_final["days_until_close"] = (
    eligible_final["close_date"] - reference_date
).dt.days

import numpy as np

def score_close_date(x):

    # Missing close date
    if pd.isna(x):
        return 0

    # More than 31 days away
    elif x > 31:
        return 15 / np.sqrt(x - 30)

    # Less than 30 days away
    elif 0 < x < 30:
        return -10 / np.sqrt(x)

    # Between 30 and 31 days
    elif 30 <= x <= 31:
        return 0

eligible_final["close_date_score"] = (
    eligible_final["days_until_close"]
    .apply(score_close_date)
)
```

```python
# Scoring: Expected number of awards
import math
def score_expected_number_of_awards(x):

    if pd.isna(x):
        return 0

    elif x <= 0:
        return 0

    else:
        return int(
            (30 / (1 + math.exp(-0.1 * x))) - 15
        )
eligible_final["expected_awards_score"] = (
    eligible_final["expected_number_of_awards"]
    .apply(score_expected_number_of_awards)
)
```

```python
# Scoring: Award average
def average_from_award_range(x):
    # -\_ graph
    scoring_max = 10
    scoring_min = -10
    if pd.isna(x) or x == 0:
        return 0
    try:
      output_score = np.clip(int((10 / (1 + math.exp(-x + 4) ) )), scoring_min, scoring_max)
    except OverflowError:
      return(scoring_max)
    return output_score
    #if x >= 0 else int(x) - 1


def award_average(ceiling, floor):
    # scores the award_average category for each grant as the average of the award_ceiling and award_floor columns
    # input: two numbers from the dataframe, award_ceiling and award_floor
    # output: the average of the two columns

    return (ceiling + floor) / 2


eligible_final["award_average"] = eligible_final.apply(
    lambda row: award_average(
        row["award_ceiling"],
        row["award_floor"]
    ),
    axis=1
)

awards_all_avg = eligible_final["award_average"].mean()

print(f"total average of award avgs: {awards_all_avg}")
print(( awards_all_avg - eligible_final["award_average"] ))

eligible_final["award_average_score"] = (
    ( eligible_final["award_average"] - awards_all_avg )
).apply(average_from_award_range)
```

```python
# Scoring: Award average

def average_from_award_range(x):
    # Convert the normalized award difference into a score from 0 to 10
    # x = 0 means the grant's award average equals the overall award average
    # Positive x means the grant's award average is above the overall average
    # Negative x means the grant's award average is below the overall average

    if pd.isna(x):
        return np.nan

    return np.clip(
        10 / (1 + np.exp(-4 * x)),
        0,
        10
    )


def award_average(ceiling, floor):
    # Calculates the average award amount for each grant
    # input: award_ceiling and award_floor
    # output: average of award_ceiling and award_floor

    return (ceiling + floor) / 2


# Calculate the average award amount for each grant
eligible_final["award_average"] = eligible_final.apply(
    lambda row: award_average(
        row["award_ceiling"],
        row["award_floor"]
    ),
    axis=1
)


# Calculate the overall average award amount across all grants
awards_all_avg = eligible_final["award_average"].mean()

print(f"total average of award avgs: {awards_all_avg}")


# Normalize each grant's award average relative to the overall average
eligible_final["award_average_normalized"] = (
    (eligible_final["award_average"] - awards_all_avg)
    / awards_all_avg
)

print(eligible_final["award_average_normalized"])


# Convert the normalized difference into a score from 0 to 10
eligible_final["award_average_score"] = (
    eligible_final["award_average_normalized"]
    .apply(average_from_award_range)
)


# Check the results
print(
    eligible_final[
        [
            "award_floor",
            "award_ceiling",
            "award_average",
            "award_average_normalized",
            "award_average_score"
        ]
    ]
)
```

```python
# Normalize each scoring criterion to 0-1
score_columns = [
    "funding_category_score",
    "category_score",
    "award_average_score",
    "expected_awards_score",
    "close_date_score"
]

for col in score_columns:
    min_score = eligible_final[col].min()
    max_score = eligible_final[col].max()

    eligible_final[col + "_normalized"] = (
        eligible_final[col] - min_score
    ) / (
        max_score - min_score
    )

# Score calculation
eligible_final["total_score"] = (
    eligible_final["funding_category_score_normalized"] * 0.65
    + eligible_final["category_score_normalized"] * 0.05
    + eligible_final["award_average_score_normalized"] * 0.15
    + eligible_final["expected_awards_score_normalized"] * 0.05
    + eligible_final["close_date_score_normalized"] * 0.10
)

# Scale to 0-100
eligible_final["total_score"] = (
    eligible_final["total_score"] * 100
)
```
